**-	Neuron, weights, bias, activation functions (ReLU, Sigmoid, Tanh)**

**Neuron**

Basic unit of a neural network.

Takes input then applies weights & bias then passes through activation function and last gives output.

Works like a mini decision-maker.

**Weights**

Numbers that show importance of each input.

Large weight = input strongly affects output.

Adjusted during training to improve accuracy.

**Bias**

Extra value added to the weighted sum.

Helps the model shift the activation function left or right.

Makes learning more flexible (not stuck at origin).

**Activation Functions**

Decide whether a neuron should (activate) or not.

Add non-linearity so the network can learn complex patterns.

**Types:**

**ReLU (Rectified Linear Unit)**

Output = 0 if input < 0, else input.

Fast, simple, widely used in deep networks.

**Sigmoid**

Output between 0 and 1.

Good for probabilities, but can cause vanishing gradients.

**Tanh (Hyperbolic Tangent)**

Output between -1 and 1.

Better than sigmoid (centered at 0), but still risk of vanishing gradients.

**-	Optimizers (SGD, Adam)**

**SGD Optimizer:**

**Optimizer:** Helps model learn by reducing loss.

**Gradient Descent:** Moves weights opposite to gradient.

**Stochastic:** Uses one sample or mini-batch instead of whole dataset.

**Fast:** Frequent updates, good for large datasets.

**Noisy path:** Zig-zag updates, but helps escape local minima.

**Learning rate sensitive:** Too high = unstable, too low = slow.

**Variants:** Mini-batch SGD, SGD with Momentum, Adam, RMSProp.

**Adam Optimizer:**

**Full form:** Adaptive Moment Estimation.

**Purpose:** Advanced optimizer for deep learning models.

**Combination:** Uses both Momentum (remembers past gradients) + RMSProp (adaptive learning rate).

**Momentum:** Smooths updates by keeping track of past directions.

**RMSProp:** Adjusts step size for each parameter individually.

**Formula idea:** Uses moving averages of gradient (m) and squared gradient (v).

**Benefits:** Fast training, stable updates, less need for tuning.

**Default choice:** Commonly used in TensorFlow and PyTorch.

**-	Forward and backward propagation  - Overfitting, underfitting, regularization**

**Forward Propagation:**

Input data: goes through layers of the network.

Each layer applies weights + activation function.

Final output (prediction) is generated.

Used during training and testing to make predictions.

**Backward Propagation:**

Compares prediction vs. actual, calculates error (loss).

Error is sent backward through the network.

Gradients (slopes) are calculated for each weight.

Weights are updated using optimizer (like SGD/Adam).

Goal: reduce error step by step.

**Overfitting:**

Model learns too much detail/noise from training data.

Performs well on training but poorly on new data.

Signs: very high training accuracy, low test accuracy.

**Underfitting:**

Model is too simple, cannot capture patterns.

Performs poorly on both training and test data.

Signs → low accuracy everywhere.

**Regularization:**

Technique to control model complexity.

Adds penalty to large weights, keeps model simple.

Prevents overfitting, improves generalization.

Types: L1 (Lasso), L2 (Ridge), Elastic Net.
