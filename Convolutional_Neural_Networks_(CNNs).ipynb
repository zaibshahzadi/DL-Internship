**Convolutional Neural Networks (CNNs)**

**Convolution:**

**Link:** https://www.geeksforgeeks.org/machine-learning/what-are-convolution-layers/

**Operation:** A filter (e.g., 3×3 matrix) moves across the input image, multiplying values and summing them to produce a feature map.

**Purpose:** Captures local spatial relationships between pixels.

**Result:** Each filter learns to detect a specific feature (edges, curves, colors, textures).

**Key Components of a Convolution Layer:**

**1. Filters (Kernels)**

Small matrices that extract specific features from the input.
**For example**, one filter might detect horizontal edges while another detects vertical edges.
The values of filters are learned and updated during training.

**2. Stride**

Refers to the step size with which the filter moves across the input data.
Larger strides result in smaller output feature maps and faster computation.

**3. Feature maps**

In Convolutional Neural Networks (CNNs), a feature map is the output you get after applying a filter (kernel) to the input data (like an image).

Each filter looks for a specific pattern (edges, corners, textures, colors).

The result of this filtering is stored in a feature map, which highlights where that pattern appears in the image.

**4. Padding**

Padding in Convolutional Neural Networks (CNNs) is a technique where we add extra pixels (usually zeros) around the edges of an image before applying convolution.

**5. Pooling**

Pooling in Convolutional Neural Networks (CNNs) is a downsampling operation used to reduce the size of feature maps while keeping the most important information.

**-	Dropout & Batch Normalization**

**Dropout**

**What it is:** A regularization technique.

**How it works:** During training, dropout randomly “turns off” (sets to zero) some neurons in a layer.

**Why:** Prevents the network from relying too much on specific neurons : reduces overfitting.

**Effect:** Forces the model to learn more robust features since it can’t depend on just a few neurons.

**Example:** If dropout rate = 0.5, then half of the neurons are ignored in each training step.

**Batch Normalization (BN)**

**What it is:** A normalization technique applied inside the network.

**How it works:** It normalizes the inputs of each layer (mean = 0, variance = 1) for every mini‑batch, then scales and shifts them with learnable parameters.

**Why:** Helps stabilize and speed up training by reducing the problem of internal covariate shift (changes in distribution of inputs during training).

**Effect:** Allows higher learning rates, faster convergence, and sometimes acts like a regularizer.

**Example:** Before activation, BN ensures values don’t explode or vanish, keeping training smooth.

**Architecture: LeNet, VGG, ResNet (intro)**

**Architecture: LeNet**

**CNN (General):**

**Input:** Image

**Convolution:** Filters extract features (edges, shapes).

**Activation (ReLU):** Adds non-linearity.

**Pooling:** Downsamples, keeps important features.

**Fully Connected:** Flatten → connect to neurons.

**Output:** Softmax → predicts class.

**LeNet (Classic CNN):**

**Input:** 32×32 grayscale image

**C1:** Convolution (6 filters → 28×28 maps)

**S2:** Pooling (average → 14×14)

**C3:** Convolution (16 filters → 10×10 maps)

**S4:** Pooling (average → 5×5)

**C5:** Convolution (acts like fully connected → 120 maps)

**F6:** Fully connected (84 neurons)

**Output:** 10 neurons (digits 0–9)

**VGG (2014):**

**Created by** Oxford’s Visual Geometry Group.

**Input:** 224×224 RGB image.

Uses small 3×3 filters stacked many times.

**Pooling:** 2×2 max pooling to reduce size.

**Depth:** VGG‑16 (16 layers) and VGG‑19 (19 layers).

**Fully Connected Layers:** 3 dense layers at the end.

**Activation:** ReLU.

**Output:** Softmax classifier , 1000 classes (ImageNet).

**Strength:** Simple design, deep layers , strong feature extraction.

**Weakness:** Very heavy (lots of parameters, needs big memory).

**ResNet (2015):**

**Problem:** Deep networks with many layers → hard to train, accuracy drops.

**Solution:** Introduced skip connections (shortcuts).

These shortcuts let information jump over layers if needed.

Allows training of very deep networks (50, 101, 152 layers).

**Strength:** Makes deep learning easier and more accurate.

**Output:** State‑of‑the‑art results, won ImageNet challenge.
